```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_csv("SpotifyFeatures.csv")
df.columns = df.columns.str.strip() 

# features to include
features = ["danceability", "energy", "valence", 
            "instrumentalness", "acousticness"]

# songs to analyze
songs_to_analyze = [
    {"track": "Bohemian Rhapsody", "artist": "Queen"},
    {"track": "Firestarter", "artist": "The Prodigy"},
    {"track": "So What", "artist": "Miles Davis"}
]

# my ratings
subjective_ratings = {
    "Bohemian Rhapsody": [0.3, 0.6, 0.6, 0.0, 0.0],
    "Firestarter": [0.9, 0.8, 0.2, 0.0, 0.0],
    "So What": [0.6, 0.2, 0.5, 0.9, 0.1]
}

# HTML sections
html_intro = """
<h2>Introduction</h2>
<p>Many often wonder how algorithms and data truly see us. Do they register everyone as being part of a community, or are we simply seen as individuals? And how does music play into that, as music has an interesting relationship with subjectivity?</p>
<p>Streaming platforms, such as Pandora, Apple Music and Spotify, utilize a numeric scale system to assign attributes to songs, thus helping with categorical grouping and personalized recommendations. However, music is experienced socially, emotionally, and personally. As Hesmondhalgh (2008) argues, music is perceived as deeply personal, and we often use it to scaffold our own identities and sense of self.</p>
<p>Therefore, this raises the question of whether numerical metrics can meaningfully capture lived listening experiences, or the quantifiability of music listening, and who ultimately gets to determine such metrics.</p>
"""

html_methodology = """
<h2>Methodology</h2>
<p>One such streaming platform that utilizes these metrics is Spotify. Spotify has numerous "hidden" features, such as danceability and valence, that it calculates using a variety of underlying metrics. These features are not readily available in the app and require access to the web page's API.</p>
<p>To analyze the value of these numeric metrics, I wanted to compare these unique "hidden" features with my own subjective and personal ratings. After much frustration trying to access Spotify’s deep-level audio features through the API, I realized that a proper license is required, which in turn requires an application. Therefore, in the following analysis, I used a pre-made dataset of 200,000 songs and associated metrics from Spotify that someone else web-scraped. The dataset and documentation can be found here: <a href='https://www.kaggle.com/datasets/zaheenhamidani/ultimate-spotify-tracks-db/data'>https://www.kaggle.com/datasets/zaheenhamidani/ultimate-spotify-tracks-db/data</a></p>
<p>The first step in my analysis was to choose three songs that reflect some of the different types of music available. I aimed to select songs of varying genres and time periods, and thus presumably different audiences. My chosen songs are the following:</p>
<ul>
<li><strong>Bohemian Rhapsody</strong> by Queen</li>
<li><strong>Firestarter</strong> by The Prodigy</li>
<li><strong>So What</strong> by Miles Davis</li>
</ul>
<p>Next, I selected my non-readily available features of interest. Below are the variables I chose, along with their definitions according to Spotify’s API documentation (<a href='https://developer.spotify.com/documentation/web-api/reference/get-audio-features'>https://developer.spotify.com/documentation/web-api/reference/get-audio-features</a>)</p>
<ul>
<li><strong>Danceability</strong> - Danceability describes how suitable a track is for dancing based on a combination of musical elements including tempo, rhythm stability, beat strength, and overall regularity. A value of 0.0 is least danceable and 1.0 is most danceable.</li>
<li><strong>Energy</strong> - Energy is a measure from 0.0 to 1.0 and represents a perceptual measure of intensity and activity. Typically, energetic tracks feel fast, loud, and noisy. For example, death metal has high energy, while a Bach prelude scores low on the scale. Perceptual features contributing to this attribute include dynamic range, perceived loudness, timbre, onset rate, and general entropy.</li>
<li><strong>Valence</strong> - A measure from 0.0 to 1.0 describing the musical positiveness conveyed by a track. Tracks with high valence sound more positive (e.g. happy, cheerful, euphoric), while tracks with low valence sound more negative (e.g. sad, depressed, angry).</li>
<li><strong>Acousticness</strong> - A confidence measure from 0.0 to 1.0 of whether the track is acoustic. 1.0 represents high confidence the track is acoustic.</li>
<li><strong>Instrumentalness</strong> - Predicts whether a track contains no vocals. "Ooh" and "aah" sounds are treated as instrumental in this context. Rap or spoken word tracks are clearly "vocal". The closer the instrumentalness value is to 1.0, the greater likelihood the track contains no vocal content. Values above 0.5 are intended to represent instrumental tracks, but confidence is higher as the value approaches 1.0.</li>
</ul>
<p>I then graphed the differences between Spotify's values for these features and my own ratings for each song on a radar plot, allowing me to visually examine where my perceptions aligned with or diverged from Spotify’s metrics. This comparison provides a foundation for discussing how well these "hidden" features capture subjective musical qualities.</p>
"""

html_analysis_conclusion = """
<h2>Analysis & Conclusion</h2>
<p>From a visual standpoint, it initially appears that my subjective ratings do not align closely with Spotify’s metrics. However, upon closer inspection, the graphs for Bohemian Rhapsody suggest that the alignment is actually fairly good overall, with my ratings generally on a slightly larger scale but showing no major discrepancies. Regarding Firestarter, I was surprised to see that Spotify does not regard it as highly danceable as I do. I considered its fast tempo and rhythmic intensity to make it very danceable. Additionally, I was shocked that Spotify rated it higher in instrumentalness, as I noticed a strong presence of vocals and layered frequencies. Lastly, what stood out to me about So What was the instrumentalness metric. Since this song has no vocals, I expected this category to be much higher.</p>
<p>It is important to note that these discrepancies may be influenced by the dataset I used, which may not perfectly reflect Spotify’s API due to my inability to access it directly.</p>
<p>While Spotify’s metrics are derived from numerous calculations and complex algorithms, it is fascinating to compare how my personal evaluations differ from what the technology determines. This raises questions about the nature of musical taste: if these metrics are arbitrarily calculated and used to recommend songs, how much autonomy do we really have over our listening choices? Are our preferences truly understood, or are they merely statistically approximated? This tension is frequently discussed in music blogs and online forums, where users critique the limits of algorithmic recommendations.</p>
<p>Reflecting on this process creates an uncomfortable dichotomy. On one hand, technological algorithms have enhanced our listening habits by offering productive recommendations and facilitating music discovery. On the other hand, the idea that our tastes are approximated through potentially imperfect metrics feels reductive. Neither perspective fully satisfies, leaving a lingering tension between subjective experience and algorithmic measurement.</p>
<p>Spotify uses these metrics to guide recommendations, such as selecting high-danceability songs for “going out” or workout playlists, or high-valence songs for happy, feel-good playlists. This creates consistency in playlists and avoids jarring transitions, enhancing cohesion and overall listening experience. In this sense, music can be viewed not only as art but also as behavioral utility, shaped to maximize enjoyment and engagement.</p>
<p>It is also interesting to consider what Spotify measures, and what it does not. For instance, there are no metrics capturing semantic meaning, or political and cultural implications in lyrics. This reflects the broader idea that sound conveys meaning not just through lyrics but through rhythm, melody, timbre, and texture, highlighting the limitations of purely numerical representations of music.</p>
<br>
<p>Sources:</p>
<p>Hesmondhalgh, David. “Towards a Critical Understanding of Music, Emotion and Self-Identity.” Consumer Markets & Culture 11, no. 4 (2008): 329–343.</p>
<p>Pelly, Liz. “The Conquest of Chill.” In Mood Machine. Atria/One Signal Publishers, 2025.</p>
<p>Prey, Robert. “Nothing Personal: Algorithmic Individuation on Music Streaming Platforms.” Media, Culture & Society 40, no. 7 (2018): 1086–1100.</p>
"""

# create html file
with open("spotify_radar_analysis.html", "w", encoding="utf-8") as f:
    f.write("<html><head><title>Spotify Metrics vs Subjective Ratings</title>")
    f.write("<style>"
        "body{font-family:Arial, sans-serif;margin:30px;background:#f9f9f9;}"
        "h1{text-align:center;font-size:32px;margin-bottom:5px;}"
        ".subtitle{text-align:center;color:#555;font-size:18px;margin-bottom:40px;}"
        ".plot{margin:50px auto;padding:20px;border:1px solid #ccc;"
        "border-radius:10px;background:#fff;box-shadow:2px 2px 10px rgba(0,0,0,0.1);}"
        "</style>")
    f.write("</head><body>")
    f.write("<h1>Subjectivity of Music Analysis</h1>")
    f.write("<div class='subtitle'>Creative Project 3 - GEN MUSIC 170 - Maya Schwartz</div>")
    f.write(html_intro)
    f.write(html_methodology)

# add radar plots to HTML
for song in songs_to_analyze:
    track_name = song["track"]
    artist_name = song["artist"]

    spotify_row = df[(df['track_name'] == track_name) & (df['artist_name'] == artist_name)]
    if spotify_row.empty:
        print(f"No data for {track_name} by {artist_name}")
        continue

    spotify_vals = spotify_row[features].iloc[0].tolist()
    spotify_vals += [spotify_vals[0]]  # close loop
    subj_vals = subjective_ratings[track_name] + [subjective_ratings[track_name][0]]

    fig = go.Figure()
    fig.add_trace(go.Scatterpolar(r=spotify_vals, theta=features + [features[0]], fill='toself', name='Spotify Metrics', line=dict(color='green')))
    fig.add_trace(go.Scatterpolar(r=subj_vals, theta=features + [features[0]], fill='toself', name='My Interpretation', line=dict(color='pink')))
    fig.update_layout(title=f"{track_name} by {artist_name}", polar=dict(radialaxis=dict(visible=True, range=[0,1])), showlegend=True)

   
    fig_html = fig.to_html(full_html=False, include_plotlyjs='cdn')
    with open("spotify_radar_analysis.html", "a", encoding="utf-8") as f:
        f.write(f"<div class='plot'>{fig_html}</div>")

# add analysis and conclusion sections
with open("spotify_radar_analysis.html", "a", encoding="utf-8") as f:
    f.write(html_analysis_conclusion)
    f.write("</body></html>")

print("Standalone interactive HTML file created: spotify_radar_analysis.html")
```
